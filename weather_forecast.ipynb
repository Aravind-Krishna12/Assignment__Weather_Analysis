```python
import pandas as pd
from pmdarima import auto_arima

# Load your data
df = pd.read_csv("your_data.csv", parse_dates=["date_column"], index_col=["date_column"])

# Resample to daily (or monthly)
df_daily = df["temperature_celsius"].resample("D").mean()

# Auto ARIMA
model = auto_arima(df_daily, seasonal=True, m=365) #m is the seasonal period.
model_fit = model.fit(df_daily)

# Forecast
forecast = model_fit.predict(n_periods=365)

#Plot the results.
import matplotlib.pyplot as plt
plt.plot(df_daily, label = "Observed")
plt.plot(forecast, label = "Forecast")
plt.legend()
plt.show()
```
